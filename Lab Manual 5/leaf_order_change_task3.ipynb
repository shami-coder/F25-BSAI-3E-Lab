```python
#MINMAX

import math

count1 = 0

def minimax(curDepth, nodeIndex, maxTurn, scores, targetDepth):

    global count1


    if curDepth == targetDepth:
        count1 += 1
        return scores[nodeIndex]

    if maxTurn:
        return max(
            minimax(curDepth + 1, nodeIndex * 2, False, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, False, scores, targetDepth)
        )

    else:
        return min(
            minimax(curDepth + 1, nodeIndex * 2, True, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, True, scores, targetDepth)
        )



scores_minmax = [5, 3, 9, 2, 5, 3, 9, 2]

treeDepth = int(math.log2(len(scores_minmax)))

count1 = 0

result1 = minimax(0, 0, True, scores_minmax, treeDepth)





#ALPH BETA PRUNING

count2 = 0

def alphabeta(depth, index, maxTurn, scores, target, alpha, beta):

    global count2
    

    if depth == target:
        count2 += 1
        return scores[index]

    if maxTurn:
        value = float('-inf')

        for i in range(2):
            value = max(value, alphabeta(
                depth + 1, index * 2 + i, False,
                scores, target, alpha, beta
            ))

            alpha = max(alpha, value)

            if alpha >= beta:
                break

        return value

    else:
        value = float('inf')

        for i in range(2):
            value = min(value, alphabeta(
                depth + 1, index * 2 + i, True,
                scores, target, alpha, beta
            ))

            beta = min(beta, value)

            if alpha >= beta:
                break

        return value



scores_alpha = [5, 3, 9, 2, 5, 3, 9, 2]

depth = int(math.log2(len(scores_alpha)))

count2 = 0

result2 = alphabeta(
    0, 0, True, scores_alpha, depth,
    float('-inf'), float('inf')
)
#RESULTS
print("====Results====\n")
print(f"Minimax Result: {result1}")
print(f"Leaf Evaluations: {count1}\n")
print("-------------------------------\n")
print(f"Alpha-Beta Result: {result2}")
print(f"Leaf Evaluations: {count2}")
```
