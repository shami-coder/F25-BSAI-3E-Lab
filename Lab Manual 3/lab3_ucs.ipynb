```python
graph = {
    'A': [('B', 3), ('C', 2)],
    'B': [('D', 5)],
    'C': [('D', 4), ('E', 2)],
    'D': [('E', 2)],
    'E': [('G', 4)],
    'G': []
}

stack = []
visited = []
total_cost = []

def ucs(start):
    stack.append(start)
    step = 1
    
    while stack:
        print(f"Step {step}:")
        print(f"  Stack       : {stack}")
        print(f"  Visited     : {visited}")
        print(f"  Cost List   : {total_cost}")
        
        
        temp = stack.pop()
        
        if temp not in visited:
            visited.append(temp)
            for child, child_cost in graph[temp]:
                stack.append(child)
                total_cost.append(child_cost)
                
        step += 1

ucs("A")


print("FINAL RESULTS:")
print("Stack       :", stack)
print("Visited     :", visited)
print("Total Cost  :", total_cost)
```
