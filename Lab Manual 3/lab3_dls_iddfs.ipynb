```python


graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["G"],
    "F": [],
    "G": []
}

def dls(node, goal, limit, path):
    if node == goal:
        return "FOUND", path
    if limit == 0:
        return "CUTOFF", None
    
    cutoff_happened = False
    
    for neighbor in graph.get(node, []):
        if neighbor in path:
            continue
            
        status, res_path = dls(neighbor, goal, limit - 1, path + [neighbor])
        
        if status == "FOUND":
            return "FOUND", res_path
        if status == "CUTOFF":
            cutoff_happened = True
            
    if cutoff_happened:
        return "CUTOFF", None
    else:
        return "FAILURE", None

def iddfs(start, goal, max_depth):
    history = []
    for limit in range(max_depth + 1):
        status, path = dls(start, goal, limit, [start])
        history.append((limit, status, path))
        
        if status == "FOUND":
            return status, path, history
            
    return status, path, history


print("=== Testing DLS (Limits 1 to 4) ===")
for limit in [1, 2, 3, 4]:
    status, path = dls("A", "G", limit, ["A"])
    print(f"Limit {limit}: Status = {status}, Path = {path}")
        
print("=== Testing IDDFS (Max Depth 4) ===")
status, path, history = iddfs("A", "G", 4)
for limit, stat, p in history:
    print(f"Limit {limit}: Status = {stat}, Path = {p}")
```
