```python


from collections import deque


graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["G"],
    "F": [],
    "G": []
}


def bfs(start, goal):
    
    queue = deque([(start, [start])])
    visited = {start}
    visit_order = []
    
    while queue:
        node, path = queue.popleft()
        visit_order.append(node)
        
        
        if node == goal:
            return path, visit_order
            
        
        for n in graph[node]:
            if n not in visited:
                visited.add(n)
                queue.append((n, path + [n]))
                
    return None, visit_order


def dfs(start, goal):
    
    stack = [(start, [start])]
    visited = set()
    visit_order = []
    
    while stack:
        node, path = stack.pop()
        
        if node == goal:
            visit_order.append(node)
            return path, visit_order
            
        if node in visited:
            continue
            
        visited.add(node)
        visit_order.append(node)
        
        
        for n in reversed(graph[node]):
            stack.append((n, path + [n]))
            
    return None, visit_order


print("--- Running BFS and DFS ---")
    
bfs_path, bfs_order = bfs("A", "G")
dfs_path, dfs_order = dfs("A", "G")

print("BFS Visit Order:", bfs_order)    
print("BFS Path:", bfs_path)

print("DFS Visit Order:", dfs_order)
print("DFS Path:", dfs_path)
```
