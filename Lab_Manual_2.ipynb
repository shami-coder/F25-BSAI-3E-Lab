```python
#in this code i used previous temarture memory

class ClassRoom:
    def __init__(self):
       self.previous_temp = None

    def act(self,temp,occupied):
        if not occupied:
            target = "ECO"
        elif temp > 26:
            target = "COOL"
        elif temp < 20:
            target = "WARM"
        else:
            target = "IDLE"

        if temp == self.previous_temp:
            command =  "NO Command Given"
        else:
            command = "Send Command"
            self.previous_temp = temp

        return target,command

Agent = ClassRoom()
Actions = [(29, True), (29, True), (20, True),
            (26, True), (19, True), (29, False)]

for step,value in enumerate(Actions,1):
    temp , occupied = value
    target, command = Agent.act(temp,occupied)
    print(f"step#{step}  value:{value}  Action Performed:{target}  Command:{command}")
    

```
