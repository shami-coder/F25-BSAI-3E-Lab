```python
import re

samples = [
    "AI, search!",
    "I can't log in.",
    "shami@gmail.com",
    "CS4944 data 2.1",
    "",
    "یہ ایک مثال ہے"
]

for text in samples:
    white_space = text.split()
    words = re.findall(r"\b\w+\b", text, flags=re.UNICODE)

    print(f"Original Text : {text}")
    print(f"white Spacing : {white_space}")
    print(f"White Space Count : {len(white_space)}\n")
    print(f"Regex Tokens : {words}")
    print(f"Regex Count : {len(words)}")
    print("----------------------------")
```
