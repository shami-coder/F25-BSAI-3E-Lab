```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


UNIVERSITY_FAQS = [
    {
        "id": 1,
        "question": "What are the university timings?",
        "answer": "Please contact the university administration for current timings.",
        "source": "University Administration"
    },
    {
        "id": 2,
        "question": "How can I check my attendance?",
        "answer": "Students can check their attendance through the university portal.",
        "source": "Student Portal"
    },
    {
        "id": 3,
        "question": "How can I contact the IT department?",
        "answer": "You can contact the university IT help desk for technical support.",
        "source": "IT Department"
    },
    {
        "id": 4,
        "question": "How can I submit assignments?",
        "answer": "Submit assignments through the university learning portal.",
        "source": "Assignment Policy"
    },
    {
        "id": 5,
        "question": "What is the minimum attendance requirement?",
        "answer": "Students must maintain at least 75 percent attendance.",
        "source": "Attendance Policy"
    }
]


questions = [faq["question"] for faq in UNIVERSITY_FAQS]

vectorizer = TfidfVectorizer()
faq_vectors = vectorizer.fit_transform(questions)


def chatbot(user_question):

    user_vector = vectorizer.transform([user_question])

    similarities = cosine_similarity(user_vector, faq_vectors)[0]

    best_index = similarities.argmax()
    best_score = similarities[best_index]

    if best_score < 0.15:
        return "Sorry, I could not find a relevant answer."

    faq = UNIVERSITY_FAQS[best_index]

    return (
        "\nAnswer: " + faq["answer"] +
        "\nSource: " + faq["source"] +
        "\nSimilarity Score: " + str(round(best_score, 3))
    )


print("University FAQ Chatbot")
print("Type 'exit' to stop.")

while True:

    user_input = input("\nYou: ")

    if user_input.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    response = chatbot(user_input)

    print("Chatbot:", response)
```
