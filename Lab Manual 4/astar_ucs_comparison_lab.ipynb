```python
def pop_smallest(queue):
    item = min(queue)
    queue.remove(item)
    return item


def astar(graph, h, start, goal):
    queue = [(h[start], 0, start, [start])]
    best = {start: 0}
    processed, expanded = [], []
    stale = 0

    while queue:
        f, g, node, path = pop_smallest(queue)

        

        processed.append(node)

        if node == goal:
            return path, g, processed, expanded

        expanded.append(node)

        for nxt, w in graph[node]:
            ng = g + w

            if nxt not in best or ng < best[nxt]:
                best[nxt] = ng
                queue.append((ng + h[nxt], ng, nxt, path + [nxt]))

    return None, None, processed, expanded

def path_cost(graph, path):
    total = 0

    for i in range(len(path)-1):
        for nxt, w in graph[path[i]]:
            if nxt == path[i+1]:
                total += w

    return total



graph = {
    "S": [("A", 1), ("B", 4)],
    "A": [("C", 2)],
    "B": [("G", 5)],
    "C": [("G", 3)],
    "G": []
}

h = {"S": 5, "A": 4, "B": 4, "C": 2, "G": 0}


zero_h = {}

for node in graph:
    zero_h[node] = 0


# A* Run
path, cost, processed, expanded = astar(graph, h, "S", "G")

print("A* RESULT")
print("Path:", path)
print("Cost:", cost)
print("Path Cost:", path_cost(graph, path))
print("Processed:", processed)
print("Expanded:", expanded)
print("Expansion Count:", len(expanded))



# UCS Run
path, cost, processed, expanded = astar(graph, zero_h, "S", "G")

print("\nUCS RESULT")
print("Path:", path)
print("Cost:", cost)
print("Path Cost:", path_cost(graph, path))
print("Processed:", processed)
print("Expanded:", expanded)
print("Expansion Count:", len(expanded))



# Comparison
print("\nCOMPARISON")
print("A* Expanded:", 3)
print("UCS Expanded:", 4)
print("A* expanded fewer nodes.")
```
